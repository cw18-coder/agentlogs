# Session logs: bundled sample

Load the repository's bundled session-log Parquet file and display its first 10 rows. This demonstration sample is not a representative research sample. No download is required.

```python
from pathlib import Path

import duckdb
from IPython.display import display
```

```python
roots = [
    path for path in [Path.cwd(), *Path.cwd().parents]
    if (path / 'packages' / 'risenlab-agentlogs').is_dir()
    and (path / 'data' / 'dataset-sample').is_dir()
]
if not roots:
    raise FileNotFoundError('Open this notebook from within the agentlogs repository.')

REPO_ROOT = roots[0]
sample_path = (
    REPO_ROOT / 'data' / 'dataset-sample' / 'agent_session_logs'
    / 'part_00000_of_00001.parquet'
)
if not sample_path.is_file():
    raise FileNotFoundError(f'Missing bundled session-log sample: {sample_path}')

con = duckdb.connect()
session_logs = con.read_parquet(str(sample_path))
```

```python
session_logs_preview = session_logs.limit(10).df()
display(session_logs_preview)
```
